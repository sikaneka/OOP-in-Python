## Inheritance
* Inheritance is the process by which one class acquires the properties and methods of another class.
* In object-oriented programming (OOP), inheritance allows a new class (called the child/subclass) to reuse and extend the features of an existing class (called the parent/superclass).

* Benefits:
    * Code reusability
    * Easy maintenance
    * Extensibility of existing code
    * Supports hierarchical classification 

<img src="Inheritance.jpg" width=500 height=200>

```python
#simple inheritance
class PersonalDetails:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def display(self):
        print(f"student name: {self.name}\nstudent age: {self.age}")

class AcademicDetails(PersonalDetails):
    def __init__(self,mark,grade,name,age):
        self.grade=grade
        self.mark=mark
        super().__init__(name,age)
    def details(self):
        print(f"This is a class for accademic details\nGrade: {self.grade}\nMark:{self.mark}")

a=AcademicDetails(30,"A","Nila",20)
a.details()
a.display()
```

```python
# Multiple inheritance
class Person:
    def __init__(self,name,age,gender):
        self.s_name=name
        self.age=age
        self.gender=gender

class Employee:
    def __init__(self,emp_id,department,salary):
        self.emp_id=emp_id
        self.dep=department
        self.salary=salary

class Teacher(Person,Employee):
    def __init__(self,name,age,gender,emp_id,dep,salary,batch):
        self.batch=batch
        super().__init__(name,age,gender)
        Employee.__init__(self,emp_id,dep,salary)
    def details(self):
        print(f"{self.age},{self.emp_id},{self.batch}")

a=Teacher("Naina",23,"F","cs_123","cs",40000,"AB")
a.details()
```

```python
# Multilevel Inheritance
class Vehicle:
    def __init__(self, clr,vehicle_no):
        self.clr=clr
        self.v_no=vehicle_no
class car(Vehicle):
    def __init__(self,clr,v_no,owner):
        self.owner=owner
        super().__init__(clr,v_no)

class Innova(car):
    def __init__(self,clr,v_no,owner,purchase_date):
        self.p_date=purchase_date
        super().__init__(clr,v_no,owner) # can also use default values instead of passing every attributes in super

    def display(self):
        print(f"{self.clr},{self.owner},{self.p_date}")

v=Vehicle('red',12354)
a=Innova('red',12354,"aswathi","12/05/25")
a.display()
```

## Polymorphism
* Polymorphism is an object-oriented programming (OOP) concept that allows the same method, function, or operator to perform different actions depending on the object that uses it.
* Benefits
    * Increases code flexibility
    * Improves reusability
    * Simplifies maintenance
    * Supports method overriding and method overloading

<b>Method Overloading</b> is a feature in which multiple methods have the same name but different parameters (different number, type, or order of parameters) within the same class.

```python
class Addition:
    def add(self,a,b):
        return a+b
    def add(self,a,b,c):
        return a+b+c
    def add(self,a,b,c,d,e):
        return a+b+c+d+e
    
ob=Addition()
ob.add(2,1,8,9,0)
```

* Python doesn't support method overloading directly, but we can achieve it through default argument method or or arbitrary argument

```python
class Addition:
    def add(self,*args):
            return sum(args)
a=Addition()
a.add(2,4)
a.add(2,5,4)
```

<b>Method overriding</b> is the process by which a child class redefines a method inherited from its parent class with the same method signature.

```python
class Animal:
    def sound(self):
        print("Animals make different sounds")

class Dog(Animal):
    def sound(self):
        print("Dog Barks")

class Cat(Animal):
    def sound(self):
        print("Cat Meows")

a=Animal()
d=Dog()
c=Cat()

a.sound()
d.sound()
c.sound()
```

## Encapsulation
* Encapsulation is the process of wrapping data and methods into a single unit (class) while restricting direct access to the data through data hiding.
* Benefits:
    * Data hiding and security
    * Better control over data access
    * Improved code maintainability
    * Increased modularity
* In Python (and OOP), encapsulation is usually demonstrated using three access levels:
    * Public
    * Protected
    * Private

<u>Public</u><br>
Public members are accessible inside and outside the class.


```python
class Student:
    def __init__(self):
        self.name = "Manu"   # Public variable

s = Student()
print(s.name)
```

<u>Protected</u>
* Members are prefixed with a single underscore (_).
* Protected members can be accessed within the class and derived classes. (By convention, they should not be accessed directly outside the class.)

```python
class Student:
    def __init__(self):
        self._name = "Manu"   # Protected variable

class Child(Student):
    def display(self):
        print(self._name)

c = Child()
c.display()
```

<u>Private</u>
* Members are restricted to the class itself and are prefixed with double underscores (__).
* Private members cannot be directly accessed outside the class and provide the highest level of data hiding.

```python
class Student:
    def __init__(self):
        self.__name = "Manu"   # Private variable

    def show(self):
        print(self.__name)

s = Student()
s.show()
```

## Abstraction
* Abstraction is an Object-Oriented Programming (OOP) concept that hides implementation details and shows only the essential features of an object to the user.
* It helps reduce complexity by allowing users to focus on what an object does rather than how it does it.
* Benefits:
    * Hides complex implementation details
    * Increases security
    * Improves code maintainability
    * Reduces complexity
    * Enhances reusability

<u>Abstract class</u>
* An Abstract Class is a class that cannot be instantiated (an object cannot be created directly from it). It is used as a blueprint for other classes and may contain both abstract methods and regular methods.
* In Python, abstract classes are created using the ABC class from the abc module.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    
    @abstractmethod
    def sound(self):
        pass
```

<u> Abstract method</u>
* An Abstract Method is a method that is declared but does not contain an implementation. Every child class must provide its own implementation of the abstract method.

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):

    def sound(self):
        print("Dog barks")

d = Dog()
d.sound()
```

```python
from abc import ABC, abstractmethod

# Abstract class
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

# Child class 1
class Car(Vehicle):
    def start(self):
        print("Car starts with a key ignition.")

    def stop(self):
        print("Car stops when brakes are applied.")

# Child class 2
class Bike(Vehicle):
    def start(self):
        print("Bike starts with a self-start button.")

    # def stop(self):
    #     print("Bike stops when brakes are applied.")

# Usage
c = Car()
b = Bike()

c.start()
c.stop()

b.start()
b.stop()
```

## Property decorator
A property decorator (@property) is a built-in Python decorator that allows a method to be accessed as an attribute, helping implement encapsulation through getter, setter, and deleter methods.

```python
class Student:
    def __init__(self):
        self.__marks = 0

    @property
    def marks(self):
        return self.__marks

    @marks.setter
    def marks(self, value):
        self.__marks = value

s = Student()

s.marks = 85      # Calls setter
print(s.marks)    # Calls getter
```
